#### 23. Create a custom dtype that describes a color as four unsigned bytes (RGBA) (★☆☆)

````python
custom_dtype = np.dtype([
    ('R', 'i8'),
    ('G', 'i8'),
    ('B', 'i8'),
    ('A', 'i8')
])

color_red = (255, 0, 0, 255)
color_green = (0, 255, 0, 255)
color_blue = (0, 0, 255, 255)

data = [
    color_red,
    color_green,
    color_blue
]

colors = np.array(data, dtype=custom_dtype)

print(colors)
````

#### 29. How to round away from zero a float array ? (★☆☆)

````python
rng = np.random.default_rng()

array = rng.random(size=(10))
rounded_array = np.ceil(np.abs(array))

result = np.copysign(array, rounded_array)
print(result)
````

#### 30. How to find common values between two arrays? (★☆☆)

````python
array1 = np.array([1, 2, 3, 4, 5])
array2 = np.array([3, 4, 5, 6, 7])

values, array1_indices, array2_indices = np.intersect1d(array1, array2, return_indices=True)

print('values: ', values)
print('indices: ', array1_indices)
````
